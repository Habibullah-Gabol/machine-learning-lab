```python
# Task # 02: 2.	Write a function is_prime(n) that returns True if n is a prime number, otherwise False.

def is_prime(x):
    count = 0
    for i in range(x):
        if(i <= 1):
            continue
        if(i > 1 & x % i == 0):
            count += 1
            
    if(count == 1):
        return True
    else:
        return False

is_prime(9)
```

```python
# Task # 04: Write a function that takes a list of numbers and returns a new list with duplicates removed, 
# preserving original order.

def remove_duplicates(numbers):
    result = []

    for num in numbers:
        if num not in result:
            result.append(num)

    return result

numbers = [1, 2, 3, 2, 4, 1, 5, 3]

print(remove_duplicates(numbers))
    
```

```python
# Task # 05: Write a program that counts the number of vowels, consonants, digits, 
# and spaces in a given string.

text = input("Enter a string: ")

vowels = 0
consonants = 0
digits = 0
spaces = 0

for char in text:
    if char in "aeiouAEIOU":
        vowels += 1
    elif char.isalpha():
        consonants += 1
    elif char.isdigit():
        digits += 1
    elif char == " ":
        spaces += 1

print(f"Vowels: {vowels} | Consonants: {consonants} | Digits: {digits} | Spaces: {spaces}")
```

```python
# Task # 06: Write a function multiply(*args) that returns the product of any number of arguments using *args.

def multiply(*args):
    product = 1

    for num in args:
        product *= num

    return product

print(multiply(2, 3, 4))
print(multiply(5, 10))
```

```python
# Task # 17: Write NumPy code to normalize an array (scale all values to the range 0-1) 
# using the formula (x - min) / (max - min).

import numpy as np

arr = np.array([10, 20, 30, 40, 50])

min_value = np.min(arr)
max_value = np.max(arr)

normalized = (arr - min_value) / (max_value - min_value)

print("Original array:", arr)
print("Normalized array:", normalized)
```

```python
# Task # 20: Create a 4x4 matrix using np.arange().reshape() and compute its transpose, 
# determinant, and inverse (if invertible).
import numpy as np

matrix = np.arange(1, 17).reshape(4, 4)

transpose = matrix.T
determinant = np.linalg.det(matrix)

print("Matrix:")
print(matrix)

print("\nTranspose:")
print(transpose)

print("\nDeterminant:", determinant)

if determinant != 0:
    inverse = np.linalg.inv(matrix)
    print("\nInverse:")
    print(inverse)
else:
    print("\nMatrix is not invertible.")
```

```python
# Task # 21: Given two NumPy 1D arrays representing vectors, compute their dot product 
# manually with a loop and verify it against np.dot().

import numpy as np

a = np.array([1, 2, 3, 4])
b = np.array([5, 6, 7, 8])

manual_dot = 0

for i in range(len(a)):
    manual_dot += a[i] * b[i]

numpy_dot = np.dot(a, b)

print("Manual dot product:", manual_dot)
print("NumPy dot product:", numpy_dot)

print("Results are equal:", manual_dot == numpy_dot)
```

```python
# Task # 24: Write a function that accepts a 2D NumPy array and returns True 
# if the array is symmetric (equal to its transpose), otherwise False.

import numpy as np

def is_symmetric(arr):
    return np.array_equal(arr, arr.T)


# Example
matrix = np.array([
    [1, 2, 3],
    [2, 4, 5],
    [3, 5, 6]
])

print(is_symmetric(matrix))
```

```python
# Task # 32:Given a DataFrame of exam scores across 3 subjects for 10 students, 
# use groupby-style aggregation to find each student's average and identify the top performer.

import pandas as pd

data = {
    "student": ["Ali", "Ahmed", "Sara", "Hassan", "Ayesha",
                "Usman", "Fatima", "Bilal", "Zain", "Hamza"],
    "Math": [85, 72, 90, 65, 88, 78, 92, 70, 82, 95],
    "Physics": [80, 75, 94, 68, 85, 80, 90, 72, 84, 91],
    "Computer": [90, 78, 96, 70, 92, 85, 95, 75, 88, 97]
}

df = pd.DataFrame(data)

average_scores = df.groupby("student")[["Math", "Physics", "Computer"]].mean()

average_scores["average"] = average_scores.mean(axis = 1)

top_student = average_scores["average"].idxmax()
top_score = average_scores["average"].max()

print("Average scores:")
print(average_scores)

print("\nTop performer:", top_student)
print("Average score:", top_score)
```

```python
# Task # 33:Given a DataFrame with a 'date' column and a 'sales' column, sort by date 
# and plot sales over time using Pandas' built-in .plot() method.
import pandas as pd

data = {
    "date": ["2026-08-05", "2026-08-01", "2026-08-04",
             "2026-08-02", "2026-08-03"],
    "sales": [500, 300, 450, 350, 400]
}

df = pd.DataFrame(data)

df["date"] = pd.to_datetime(df["date"])

df = df.sort_values("date")

df.plot(x = "date", y = "sales", kind = "line", title = "Sales Over Time",
        xlabel = "Date",
        ylabel = "Sales",
        marker = "o")

import matplotlib.pyplot as plt
plt.show()
```

```python
# Task # 34: Given a DataFrame with missing values in a 'score' column, fill the missing values 
# with the column mean and confirm no NaNs remain.

import pandas as pd

df = pd.DataFrame({
    "name": ["Ali", "Ahmed", "Sara", "Hassan", "Ayesha"],
    "score": [80, 75, None, 90, None]
})

mean_score = df["score"].mean()

df["score"] = df["score"].fillna(mean_score)

print(df)

print("\nNaNs remaining:", df["score"].isna().sum())
```
